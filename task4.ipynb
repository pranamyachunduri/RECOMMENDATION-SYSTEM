```python
# RECOMMENDATION SYSTEM USING COLLABORATIVE FILTERING

# Step 1: Import Libraries
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

# Step 2: Create Sample User-Item Rating Dataset
data = {
    'Movie1': [5, 4, 0, 1, 0],
    'Movie2': [4, 0, 0, 1, 2],
    'Movie3': [1, 1, 0, 5, 4],
    'Movie4': [0, 0, 5, 4, 5],
    'Movie5': [0, 1, 5, 4, 0]
}

# User names
users = ['User1', 'User2', 'User3', 'User4', 'User5']

# Create DataFrame
df = pd.DataFrame(data, index=users)

print("User-Movie Rating Matrix:\n")
print(df)

# Step 3: Calculate Similarity Between Users
similarity = cosine_similarity(df)

# Convert similarity matrix to DataFrame
similarity_df = pd.DataFrame(
    similarity,
    index=users,
    columns=users
)

print("\nUser Similarity Matrix:\n")
print(similarity_df)

# Step 4: Recommend Movies for User1
target_user = 'User1'

# Find most similar user
similar_user = similarity_df[target_user].sort_values(
    ascending=False
).index[1]

print("\nMost Similar User to", target_user, ":", similar_user)

# Get movies not watched by target user
recommendations = []

for movie in df.columns:
    
    if df.loc[target_user, movie] == 0 and \
       df.loc[similar_user, movie] >= 4:
        
        recommendations.append(movie)

# Step 5: Display Recommendations
print("\nRecommended Movies for", target_user, ":")

if recommendations:
    for movie in recommendations:
        print(movie)
else:
    print("No recommendations found")
```
